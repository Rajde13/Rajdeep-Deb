***1. Libraries***

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, MinMaxScaler

%matplotlib inline
```

**2. Dataset**

```python
import pandas as pd

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Fishing.csv"

df = pd.read_csv(url)

df.head()
```

**3. Applying basic statistical analysis**

```python
print("Shape:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.describe(include='object')
```

```python
df.isnull().sum()
```

**4. Handling missing data**

```python

df = df.fillna(df.mean(numeric_only=True))

for col in df.select_dtypes(include=['object']):
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```

```python
df.isnull().sum()
```

**5. Data Cleaning**

```python
df.drop_duplicates(inplace=True)
```

```python
print(df.duplicated().sum())
```

```python
df.columns=df.columns.str.strip()
```

**6. Data Transformation**

Encode the target column.

```python
encoder=LabelEncoder()

df["mode"]=encoder.fit_transform(df["mode"])
```

Normalize numerical columns.

```python
scaler=MinMaxScaler()

cols=df.columns.drop("mode")

df[cols]=scaler.fit_transform(df[cols])
```

Display transformed data.

```python
df.head()
```

**7. UNIVARIATE ANALYSIS**

Histogram

```python
plt.figure(figsize=(7,5))
sns.histplot(df["income"],bins=20,kde=True)
plt.title("Income Distribution")
plt.show()
```

Boxplot

```python
plt.figure(figsize=(7,5))
sns.boxplot(y=df["price"])
plt.title("Price Boxplot")
plt.show()
```

Countplot

```python
plt.figure(figsize=(7,5))
sns.countplot(x=df["mode"])
plt.title("Fishing Mode Count")
plt.show()
```

**8. BIVARIATE ANALYSIS**

Scatter Plot

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x="income",y="price",hue="mode",data=df)
plt.show()
```

Bar Plot

```python
plt.figure(figsize=(7,5))
sns.barplot(x="mode",y="catch",data=df)
plt.show()
```

Boxplot

```python
plt.figure(figsize=(7,5))
sns.boxplot(x="mode",y="income",data=df)
plt.show()
```

**9. MULTIVARIATE ANALYSIS**

Pair Plot

```python
sns.pairplot(df[["price","income","catch","mode"]],hue="mode")
plt.show()
```

Correlation Heatmap

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(),annot=True,cmap="coolwarm")
plt.show()
```

Bubble Plot

```python
plt.figure(figsize=(8,6))

plt.scatter(df["price"],
            df["catch"],
            s=df["income"]*300,
            c=df["mode"],
            alpha=0.6)

plt.xlabel("Price")
plt.ylabel("Catch")
plt.title("Price vs Catch (Bubble Size = Income)")
plt.show()
```
